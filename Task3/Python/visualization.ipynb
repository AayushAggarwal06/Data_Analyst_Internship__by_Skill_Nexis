# Data Visualization

# Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Load the Dataset

```python
df = pd.read_csv("../Results/cleaned_data.csv")
```

# Display the Dataset

```python
df.head()
```

# Bar Chart - Average Calories by Workout Intensity

```python
avg_calories = df.groupby("Workout_Intensity")["Calories"].mean()

plt.figure(figsize=(6,4))
avg_calories.plot(kind="bar")

plt.title("Average Calories by Workout Intensity")
plt.xlabel("Workout Intensity")
plt.ylabel("Average Calories")

plt.tight_layout()
plt.show()
```

# Line Chart - Calories Burned Over Time

```python
plt.figure(figsize=(10,5))

plt.plot(df["Date"], df["Calories"], marker="o")

plt.title("Calories Burned Over Time")
plt.xlabel("Date")
plt.ylabel("Calories")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

# Histogram - Pulse Distribution

```python
plt.figure(figsize=(7,4))

plt.hist(df["Pulse"], bins=8)

plt.title("Pulse Distribution")
plt.xlabel("Pulse")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

# Scatter Plot - Pulse vs Calories

```python
plt.figure(figsize=(7,5))

plt.scatter(df["Pulse"], df["Calories"])

plt.title("Pulse vs Calories Burned")
plt.xlabel("Pulse")
plt.ylabel("Calories")

plt.tight_layout()
plt.show()
```

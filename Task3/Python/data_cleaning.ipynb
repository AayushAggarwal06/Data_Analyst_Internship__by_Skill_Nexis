# Remove Duplicate Rows

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

# Rows with Missing Date

```python
df[df["Date"].isnull()]
```

```python
df = df.dropna(subset=["Date"])
```

# Fill Missing Calories

```python
df["Calories"].median()
```

```python
df["Calories"] = df["Calories"].fillna(df["Calories"].median())
```

```python
df.isnull().sum()
```

```python
df.shape
```

# Verifying Data Types

```python
df.dtypes
```

```python
df["Date"]
```

# Convert Date Column to Datetime

```python
df["Date"] = pd.to_datetime(df["Date"], format="mixed")
```

```python
df.dtypes
```

# Feature Engineering

## 1. Calories Burned Per Minute

```python
df["Calories_per_Minute"] = (df["Calories"] / df["Duration"]).round(2)
```

```python
df['Calories_per_Minute']
```

## 2. Pulse Difference

```python
df["Pulse_Difference"] = df["Maxpulse"] - df["Pulse"]
```

```python
df['Pulse_Difference']
```

## 3. Workout Intensity

```python
df["Workout_Intensity"] = df["Pulse"].apply(
    lambda x: "High" if x >= 120 else "Normal"
)
```

```python
df['Workout_Intensity']
```
